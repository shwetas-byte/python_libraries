# SERIES

```python
import numpy as np
import pandas as pd
```

```python
a=pd.Series([12,45,67])
print(a)
print(type(a))
```

### How to create series using numpy array

```python
arr1=np.array([1,2,3,4,5,6,7])
print(arr1.ndim)
print(type(arr1))
```

```python
s=pd.Series(arr1)
print(type(s))
print(s)
```

### How to create series from dictionary

```python
di={"marks":[20,30,40,50,60]}
```

```python
s=pd.Series(di)
print(s)
print(type(s))
```

se=pd.Series([10,12,34,56])

### How to give name to series

```python
se=pd.Series([12,4,34,56,7,8,65],name='markss')
print(se)
```

### How to give custom indexing

```python
se=pd.Series([12,4,34,56,7,8,65],index=['a','b','c','d','e','f','g'])
print(se)
```

```python
s=pd.Series([56,32,97,84,71],index=['shweta','pihu','pranjal','riya','ayushi'])
print(s)
```

### How to create dataframe from dictionary

```python
d={
    'eng':[14,34,32,26],
    'hindi':[34,66,77,89],
    'Python':[23,21,34,54]
}
df=pd.DataFrame(d,index=['shweta','pihu','pranjal','riya'])
# print(df)
display(df)
```

```python
d={
    'id':[np.nan,102,103,104,np.nan],
    'name':['AA','BB',np.nan,'CC',np.nan],
    'age':[23,np.nan,12,np.nan,67]
}
df=pd.DataFrame(d)
df
```

### to find misiing value in pandas

```python
display(df.isna())   ## it tells true where nan values found and other false
```

```python
display(df.isna().sum())  # it tells total nan values for each column
```

```python
display(df.isna().sum().sum())  ## it tells total no. of nan values of whole dataset
```
