### 2.)Create a 3x3 array of all zeros, then a 3x3 identity matrix.


```python
zero=np.zeros((3,3),dtype=int)
print("3x3 Zero matrix is:")
print(zero)
identy = np.eye(3,dtype=int)
print("3x3 Identity matrix is:")
print(identy)
```

### 3.)Create np.arange(20), reshape it into a 4x5 array, and print its .ndim, .shape, and .size.


```python
arr = np.arange(20).reshape(4,5)
print("Array is:")
print(arr)
print(arr.ndim)
print(arr.shape)
print(arr.size)
```

### 13.) Create a (2,3) array and produce both its .flatten() and .ravel() versions, then modify the raveled version's first element and print the original array.


```python
ar=np.random.randint(12,50,(2,3))
print(ar)
rav=ar.ravel()
rav[0]=2345
print("Original Array:")
print(ar)
print("Array after ravel:")
print(rav)
```
