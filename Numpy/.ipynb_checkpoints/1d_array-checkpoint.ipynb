## Array attribute**
#### jo cheej parameter accept kre vo function hote haii

#### 1) .ndim -- to check dimension of array and it is a attribue

```python
arr = np.array(a)
print(arr)
print(arr.ndim)    # it shows the dimension of array
```

#### 2)  .shape -- it shows the number of rows and column

```python
arr = np.array(a)
print(arr)
print(arr.shape)
```

#### 3) .size--- it shows total no. of elemnt in the array

```python
arr = np.array(a)
print(arr)
print(arr.size)
```

#### 4) .dtype ---  it shows the datatype of  elements  inside array 
#### if there is different datatype of elements in a array it converts whole array in the preferred datatype acc. to their preference

```python
# string ka dtype --- <U49 ---- here 49 is string memory length and it changes usually and depends on different system--- u is unicode
arr1 = [12,13,14]
arr2=[12,13.7,18]
arr3=[12,17.8,'code']

arr1=np.array(arr1)
print(arr1)
print(arr1.dtype)
print(type(arr1))

arr2=np.array(arr2)
print(arr2)
print(arr2.dtype)
print(type(arr2))

arr3=np.array(arr3)
print(arr3)
print(arr3.dtype)
print(type(arr3))
```

#### 5) .itemsize -- single element ka dataype kitni memory store kra hai vo btata hai

```python
str_array = np.array(['12','34',45])
float_array = np.array([12.45,34,45])
int_array = np.array([12,17,20])

print(str_array)
print(str_array.itemsize)
print(float_array)
print(float_array.itemsize)
print(int_array)
print(int_array.itemsize)
```

```python
arr1=np.array([1,2,3,4,5])
print((arr1.itemsize)*arr1.size)
```
