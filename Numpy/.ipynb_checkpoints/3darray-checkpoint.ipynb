```python
import numpy as np
```

```python
# arr=np.random.random((2,3))
print(arr)
```

```python
arr1=np.random.randint(10,50,(2,3))
print(arr1)
```

```python
arr2=np.random.randint(2,30,12).reshape(3,4)
print(arr2)
```

```python
arr3=np.array([[10,20],[20,30],[30,40]])
print(arr3)
```

```python
arr4=np.arange(1,10).reshape(3,3)
print(arr4)
```

```python
arr5=np.ones((3,3),dtype=int)
print(arr5)
```

```python
arr6=np.zeros((3,3),dtype=int)
print(arr6)
```

```python
arr7=np.eye(3,dtype=int)
print(arr7)
```

```python
arr8=np.full((3,3),30)
print(arr8)
```

```python
arr9=np.linspace(5,20,9,dtype=int).reshape(3,3)
print(arr9)
```

### syntax:(depth/template,row,column)
